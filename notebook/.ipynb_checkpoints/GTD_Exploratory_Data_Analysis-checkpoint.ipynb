# 1 Load the Analytical Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

project_path = r"C:\Portfolio Projects\Global Terrorism Analysis"

cleaned_path = os.path.join(
    project_path,
    "data",
    "cleaned"
)

file_path = os.path.join(
    cleaned_path,
    "global_terrorism_analysis.csv"
)

df_analysis = pd.read_csv(file_path, low_memory=False)

print("Dataset loaded successfully!")
print("Shape:", df_analysis.shape)

df_analysis.head()
```

```python
visualization_path = os.path.join(
    project_path,
    "visualizations"
)

os.makedirs(visualization_path, exist_ok=True)

print("Visualization folder ready:")
print(visualization_path)
```

# 2 Analyze Global Incident Trends by Year
How did the number of recorded terrorist incidents change over time?

### 2.1 Caluclate Incidents per Year

```python
yearly_incidents = (
    df_analysis
    .groupby("iyear")["eventid"]
    .nunique()
    .reset_index(name="Total_Incidents")
    .sort_values("iyear")
)

yearly_incidents.head()
```

### 2.2 Identify the Years with the Highest Incidents Counts

```python
top_10_years = yearly_incidents.nlargest(
    10,
    "Total_Incidents"
)

top_10_years
```

### 2.3 Visualize the Trend

```python
plt.figure(figsize=(16, 8))

plt.plot(
    yearly_incidents["iyear"],
    yearly_incidents["Total_Incidents"],
    marker="o",
    markersize=3
)

plt.title("Global Terrorism Incidents by Year")
plt.xlabel("Year")
plt.ylabel("Number of Recorded Incidents")
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_path,
        "incidents_by_year.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

# 7 Calculate Your First Analytical KP1s

```python
total_incidents = df_analysis["eventid"].nunique()

total_countries = df_analysis["country_txt"].nunique()

total_regions = df_analysis["region_txt"].nunique()

total_recorded_fatalities = df_analysis["nkill"].sum(
    min_count=1
)

total_recorded_injuries = df_analysis["nwound"].sum(
    min_count=1
)

known_fatality_records = df_analysis["nkill"].notna().sum()

known_injury_records = df_analysis["nwound"].notna().sum()

print("=" * 50)
print("GLOBAL TERRORISM ANALYSIS — KPI SUMMARY")
print("=" * 50)

print(f"Total Recorded Incidents: {total_incidents:,}")
print(f"Countries Represented: {total_countries:,}")
print(f"Regions Represented: {total_regions:,}")

print(
    f"Recorded Fatalities (known counts): "
    f"{total_recorded_fatalities:,.0f}"
)

print(
    f"Recorded Injuries (known counts): "
    f"{total_recorded_injuries:,.0f}"
)

print(
    f"Incidents with Known Fatality Counts: "
    f"{known_fatality_records:,}"
)

print(
    f"Incidents with Known Injury Counts: "
    f"{known_injury_records:,}"
)

print("=" * 50)
```

# 8 Advanced Analysis 1 - Year Over Year Growth

### 8.1.1 Calculate Annual Growth

```python
# Year-over-year incident analysis

yearly_incidents = yearly_incidents.sort_values("iyear").copy()

yearly_incidents["Previous_Year_Incidents"] = (
    yearly_incidents["Total_Incidents"].shift(1)
)

yearly_incidents["YoY_Change"] = (
    yearly_incidents["Total_Incidents"]
    - yearly_incidents["Previous_Year_Incidents"]
)

yearly_incidents["YoY_Growth_Percentage"] = (
    yearly_incidents["YoY_Change"]
    / yearly_incidents["Previous_Year_Incidents"]
) * 100

yearly_incidents.tail(15)
```

```python
integer_columns = [
    "iyear",
    "Total_Incidents",
    "Previous_Year_Incidents",
    "YoY_Change"
]

yearly_incidents[integer_columns] = (
    yearly_incidents[integer_columns].astype("Int64")
)
```

### 8.1.2 Find The Largest Annual Increase and Decrease

```python
# Exclude the first year because it has no previous year

valid_growth = yearly_incidents.dropna(
    subset=["YoY_Growth_Percentage"]
)

largest_increase = valid_growth.loc[
    valid_growth["YoY_Change"].idxmax()
]

largest_decrease = valid_growth.loc[
    valid_growth["YoY_Change"].idxmin()
]

print("Largest annual increase:")
print(largest_increase)

print("\nLargest annual decrease:")
print(largest_decrease)
```

# 8.4 Advanced Analysis 4 — Monthly Distribution

```python
monthly_incidents = (
    df_analysis
    .groupby("imonth")["eventid"]
    .nunique()
    .reset_index(name="Total_Incidents")
    .sort_values("imonth")
)

monthly_incidents
```

```python
month_names = {
    0: "Unknown",
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December"
}

monthly_incidents["Month_Name"] = (
    monthly_incidents["imonth"].map(month_names)
)

monthly_incidents
```

```python
known_months = monthly_incidents[
    monthly_incidents["imonth"].between(1, 12)
]

plt.figure(figsize=(12, 6))

plt.bar(
    known_months["Month_Name"],
    known_months["Total_Incidents"]
)

plt.title("Recorded Incidents by Month")
plt.xlabel("Month")
plt.ylabel("Number of Recorded Incidents")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    os.path.join(
        visualization_path,
        "Incidents_by_Month.png"
    ),
    dpi=300,
    bbox_inches="tight"
)


plt.show()
```

### 8.4.1 average recorded fatalities and injuries per incident by attack type.

```python
attack_severity = (
    df_analysis.groupby("attacktype1_txt")
    .agg(
        Total_Incidents=("eventid", "nunique"),
        Known_Fatality_Records=("nkill", "count"),
        Recorded_Fatalities=("nkill", "sum"),
        Known_Injury_Records=("nwound", "count"),
        Recorded_Injuries=("nwound", "sum")
    )
    .reset_index()
)

# Calculate averages using records with known casualty counts
attack_severity["Avg_Fatalities_Per_Known_Record"] = (
    attack_severity["Recorded_Fatalities"]
    / attack_severity["Known_Fatality_Records"]
)

attack_severity["Avg_Injuries_Per_Known_Record"] = (
    attack_severity["Recorded_Injuries"]
    / attack_severity["Known_Injury_Records"]
)

attack_severity = attack_severity.sort_values(
    "Avg_Fatalities_Per_Known_Record",
    ascending=False
)

attack_severity.round(2)
```
