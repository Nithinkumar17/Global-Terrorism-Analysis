```python
import pandas as pd
import numpy as np
import os
import glob
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
# Set the project folder path
project_path = r"C:\Portfolio Projects\Global Terrorism Analysis"

# Define the raw data folder
raw_path = os.path.join(project_path, "data", "rawdata")

# Check whether the folder exists
if os.path.exists(raw_path):
    print("Raw data folder found!")
    
    files = os.listdir(raw_path)
    
    print("\nFiles available:")
    
    for file in files:
        print(file)
else:
    print("Folder not found. Check your project path.")
```

```python
# Find Excel files in the raw folder
excel_files = glob.glob(os.path.join(raw_path, "*.xlsx"))

print("Excel files found:", excel_files)
```

```python
file_path = excel_files[0]

df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
```

```python
# Dataset overview

print("=" * 50)
print("GLOBAL TERRORISM DATASET - INITIAL PROFILE")
print("=" * 50)

print(f"Total Rows: {df.shape[0]:,}")
print(f"Total Columns: {df.shape[1]:,}")
print(f"Memory Usage: {df.memory_usage(deep=True).sum() / (1024**2):.2f} MB")
print(f"Duplicate Rows: {df.duplicated().sum():,}")
print(f"Unique Event IDs: {df['eventid'].nunique():,}")

print("=" * 50)
```

```python
# Missing value analysis

missing_analysis = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df)) * 100,
    "Data_Type": df.dtypes.astype(str)
})

missing_analysis = missing_analysis.sort_values(
    by="Missing_Percentage",
    ascending=False
)

missing_analysis.head(30)
```

```python
# Check duplicate records

duplicate_rows = df.duplicated().sum()

duplicate_eventids = df["eventid"].duplicated().sum()

print("Exact duplicate rows:", duplicate_rows)
print("Duplicate event IDs:", duplicate_eventids)
print("Unique event IDs:", df["eventid"].nunique())
```

```python
# Important columns for analysis

important_columns = [
    "eventid",
    "iyear",
    "imonth",
    "iday",
    "country_txt",
    "region_txt",
    "city",
    "attacktype1_txt",
    "targtype1_txt",
    "nkill",
    "nwound"
]

available_columns = [
    col for col in important_columns
    if col in df.columns
]

important_missing = pd.DataFrame({
    "Missing_Count": df[available_columns].isnull().sum(),
    "Missing_Percentage": (
        df[available_columns].isnull().sum() / len(df)
    ) * 100
})

important_missing
```

```python
# Inspect date fields

print("Year distribution:")
print(df["iyear"].value_counts().sort_index().head(10))

print("\nMonth distribution:")
print(df["imonth"].value_counts().sort_index())

print("\nDay distribution:")
print(df["iday"].value_counts().sort_index())
```

```python
# Validate date components

invalid_year = (
    df["iyear"].isna() |
    ~df["iyear"].between(1970, 2020)
)

invalid_month = (
    df["imonth"].notna() &
    ~df["imonth"].between(0, 12)
)

invalid_day = (
    df["iday"].notna() &
    ~df["iday"].between(0, 31)
)

print("Years outside the expected range:", invalid_year.sum())
print("Invalid month values:", invalid_month.sum())
print("Invalid day values:", invalid_day.sum())
```

```python
# Casualty data quality assessment

casualty_columns = ["nkill", "nwound"]

for col in casualty_columns:
    print(f"\n--- {col} ---")
    
    print("Missing values:", df[col].isna().sum())
    print("Negative values:", (df[col] < 0).sum())
    print("Zero values:", (df[col] == 0).sum())
    print("Maximum value:", df[col].max())
    
    print("\nDescriptive statistics:")
    print(df[col].describe())
```

```python
# Phase 3: Initial data-quality report

quality_report = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Columns",
        "Exact Duplicate Rows",
        "Duplicate Event IDs",
        "Unique Event IDs",
        "Missing City",
        "Missing Fatalities",
        "Missing Injuries",
        "Negative Fatality Values",
        "Negative Injury Values"
    ],
    "Value": [
        len(df),
        len(df.columns),
        df.duplicated().sum(),
        df["eventid"].duplicated().sum(),
        df["eventid"].nunique(),
        df["city"].isna().sum(),
        df["nkill"].isna().sum(),
        df["nwound"].isna().sum(),
        (df["nkill"] < 0).sum(),
        (df["nwound"] < 0).sum()
    ]
})

quality_report
```

```python
# Check unknown month and day values

unknown_month = (df["imonth"] == 0).sum()
unknown_day = (df["iday"] == 0).sum()

unknown_month_and_day = (
    (df["imonth"] == 0) &
    (df["iday"] == 0)
).sum()

unknown_day_with_known_month = (
    (df["imonth"] != 0) &
    (df["iday"] == 0)
).sum()

print("Incidents with unknown month:", unknown_month)
print("Incidents with unknown day:", unknown_day)

print(
    "Incidents with both month and day unknown:",
    unknown_month_and_day
)

print(
    "Incidents with known month but unknown day:",
    unknown_day_with_known_month
)
```

```python
print("Month distribution:")
print(df["imonth"].value_counts().sort_index())

print("\nDay distribution:")
print(df["iday"].value_counts().sort_index())
```

```python
# Create a casualty completeness report

casualty_report = pd.DataFrame({
    "Metric": [
        "Total Incidents",
        "Known Fatality Counts",
        "Missing Fatality Counts",
        "Known Injury Counts",
        "Missing Injury Counts",
        "Both Fatalities and Injuries Missing",
        "Fatalities Known but Injuries Missing",
        "Injuries Known but Fatalities Missing"
    ],
    "Count": [
        len(df),
        df["nkill"].notna().sum(),
        df["nkill"].isna().sum(),
        df["nwound"].notna().sum(),
        df["nwound"].isna().sum(),
        (df["nkill"].isna() & df["nwound"].isna()).sum(),
        (df["nkill"].notna() & df["nwound"].isna()).sum(),
        (df["nwound"].notna() & df["nkill"].isna()).sum()
    ]
})

casualty_report
```

```python
# Inspect the 10 highest fatality counts

fatality_columns = [
    "eventid",
    "iyear",
    "imonth",
    "iday",
    "country_txt",
    "region_txt",
    "city",
    "attacktype1_txt",
    "nkill",
    "nwound"
]

fatality_columns = [
    col for col in fatality_columns
    if col in df.columns
]

df.nlargest(
    10,
    "nkill"
)[fatality_columns]
```

```python
# Select columns for the analytical dataset

selected_columns = [
    "eventid",
    "iyear",
    "imonth",
    "iday",
    "country",
    "country_txt",
    "region",
    "region_txt",
    "provstate",
    "city",
    "latitude",
    "longitude",
    "success",
    "suicide",
    "attacktype1",
    "attacktype1_txt",
    "targtype1",
    "targtype1_txt",
    "targsubtype1",
    "targsubtype1_txt",
    "gname",
    "weaptype1",
    "weaptype1_txt",
    "nkill",
    "nwound",
    "property",
    "propextent",
    "propextent_txt",
    "ishostkid",
    "INT_LOG",
    "INT_IDEO",
    "INT_MISC",
    "INT_ANY"
]

# Retain only columns that exist in this dataset
selected_columns = [
    col for col in selected_columns
    if col in df.columns
]

df_analysis = df[selected_columns].copy()

print("Original dataset shape:", df.shape)
print("Analytical dataset shape:", df_analysis.shape)

df_analysis.head()
```

```python
# Validate the analytical dataset

print("=" * 45)
print("ANALYTICAL DATASET VALIDATION")
print("=" * 45)

print("Rows:", len(df_analysis))
print("Columns:", len(df_analysis.columns))

print(
    "Duplicate event IDs:",
    df_analysis["eventid"].duplicated().sum()
)

print(
    "Missing event IDs:",
    df_analysis["eventid"].isna().sum()
)

print(
    "Missing country names:",
    df_analysis["country_txt"].isna().sum()
)

print(
    "Missing region names:",
    df_analysis["region_txt"].isna().sum()
)

print("=" * 45)
```

```python
cleaned_path = os.path.join(
    project_path,
    "data",
    "cleaned"
)

os.makedirs(cleaned_path, exist_ok=True)

output_file = os.path.join(
    cleaned_path,
    "global_terrorism_analysis.csv"
)

df_analysis.to_csv(
    output_file,
    index=False,
    encoding="utf-8"
)

print("Analytical dataset saved successfully!")
print("Location:", output_file)
print("Rows saved:", len(df_analysis))
print("Columns saved:", len(df_analysis.columns))
```
